Does the pattern holds within the same road type?

```python
import numpy as np
import matplotlib.pyplot as plt

factor = 'weather'      # the pattern you're checking
partner = 'street_lighting'   # second factor for the interaction check (Q3)
MIN_N = 30              # groups smaller than this are weak evidence

def heatmap(ax, rates, counts, title, cmap='Reds'):
    im = ax.imshow(rates.values, cmap=cmap, vmin=0, vmax=max(rates.max().max(), 0.01))
    ax.set_xticks(range(rates.shape[1]), rates.columns, rotation=30, ha='right')
    ax.set_yticks(range(rates.shape[0]), rates.index)
    for i in range(rates.shape[0]):
        for j in range(rates.shape[1]):
            r, n = rates.iat[i, j], counts.iat[i, j]
            if np.isnan(r):
                continue
            weak = n < MIN_N
            ax.text(j, i, f"{r:.0%}\nn={int(n)}", ha='center', va='center', fontsize=8,
                    color='grey' if weak else 'black', style='italic' if weak else 'normal')
    ax.set_title(title)
    plt.colorbar(im, ax=ax, label='Problem rate')
```

Useful for action, or merely interesting?

```python
# Q4 - Useful for action, or merely interesting?
levers = {   # factor -> what an operator could actually do about it
    'weather': 'Avoid / delay trip', 'time_of_day': 'Schedule trip',
    'road_type': 'Reroute', 'traffic_level': 'Reroute / reschedule',
    'road_surface': 'Reroute', 'junction_complexity': 'Reroute',
    'street_lighting': 'Reroute / infrastructure', 'digital_infrastructure': 'Infrastructure upgrade',
    'road_markings': 'Infrastructure upgrade', 'sensor_suite': 'Assign better vehicle',
    'vehicle_model': 'Assign better vehicle', 'city': 'No direct lever'}
rows = []
for col, lever in levers.items():
    g = df.groupby(col)['problem'].agg(['mean', 'count'])
    g = g[g['count'] >= MIN_N]               # ignore weak groups
    if len(g) > 1:
        rows.append((col, g['mean'].max() - g['mean'].min(), lever))
gap = pd.DataFrame(rows, columns=['factor', 'gap', 'lever']).sort_values('gap')
colors = ['tab:grey' if l == 'No direct lever' else 'tab:green' for l in gap['lever']]
fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(gap['factor'], gap['gap'], color=colors)
for y, (g, l) in enumerate(zip(gap['gap'], gap['lever'])):
    ax.text(g, y, f'  {l}', va='center', fontsize=8)
ax.set_xlabel('Gap between best and worst group (problem rate, groups >= 30 journeys)')
ax.set_title('Q4: Which patterns are actionable?\nLong green bar = big effect with a lever to pull')
plt.tight_layout(); plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

overall = df['problem'].mean()

def rate_table(col):
    return df.groupby(col, observed=True)['problem'].agg(rate='mean', n='count')

def bar(ax, col, title):
    t = rate_table(col)
    bars = ax.bar(t.index.astype(str), t['rate'], color='tab:blue')
    ax.axhline(overall, ls='--', c='k', lw=1, label=f'Overall {overall:.1%}')
    for b, r, n in zip(bars, t['rate'], t['n']):
        ax.text(b.get_x() + b.get_width()/2, r, f'{r:.1%}\nn={n}', ha='center', va='bottom', fontsize=8)
    ax.set_title(title); ax.set_ylabel('Problem rate'); ax.legend(fontsize=8)
    ax.tick_params(axis='x', rotation=30); ax.set_ylim(0, t['rate'].max() * 1.3)

def say(col):
    t = rate_table(col).sort_values('rate')
    lo, hi = t.iloc[0], t.iloc[-1]
    print(f"{col}: '{t.index[-1]}' journeys have a {hi.rate:.1%} problem rate (n={hi.n}); "
          f"'{t.index[0]}' journeys have {lo.rate:.1%} (n={lo.n}).")
```

Do difficult condictions combine

```python
# A level counts as "difficult" if its problem rate is above the overall rate
for col in ['weather', 'traffic_level', 'road_surface']:
    hard = rate_table(col).query('rate > @overall').index
    df[f'hard_{col}'] = df[col].isin(hard).astype(int)
    say(col)
df['hard_visibility'] = (df['visibility_km'] < df['visibility_km'].quantile(0.25)).astype(int)
df['visibility_band'] = pd.qcut(df['visibility_km'], 4, labels=['Lowest 25%', 'Low', 'High', 'Highest 25%'])
df['difficult_count'] = df[['hard_weather', 'hard_traffic_level', 'hard_road_surface', 'hard_visibility']].sum(axis=1)
say('visibility_band'); say('difficult_count')

fig, axes = plt.subplots(1, 3, figsize=(18, 4.5))
bar(axes[0], 'difficult_count', 'Problem rate by number of difficult conditions')
axes[0].set_xlabel('Difficult conditions present (0-4)')
bar(axes[1], 'visibility_band', 'Problem rate by visibility')
heat = df.pivot_table(index='weather', columns='traffic_level', values='problem', aggfunc='mean')
im = axes[2].imshow(heat, cmap='Reds'); plt.colorbar(im, ax=axes[2], label='Problem rate')
axes[2].set_xticks(range(heat.shape[1]), heat.columns, rotation=30); axes[2].set_yticks(range(heat.shape[0]), heat.index)
for i in range(heat.shape[0]):
    for j in range(heat.shape[1]):
        axes[2].text(j, i, f'{heat.iat[i, j]:.0%}', ha='center', va='center', fontsize=8)
axes[2].set_title('Weather x traffic')
plt.tight_layout(); plt.show()
```

Prediction model

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             classification_report, ConfusionMatrixDisplay)

df['uneven_surface'] = (df['road_surface'] == 'Uneven').astype(int)

# Only information known BEFORE the journey starts (no alerts, delays, interventions, etc.)
cat_cols = ['weather', 'traffic_level', 'time_of_day', 'road_type', 'junction_complexity',
            'road_markings', 'street_lighting', 'digital_infrastructure', 'sensor_suite']
num_cols = ['visibility_km', 'distance_km', 'uneven_surface', 'software_version',
            'vehicle_age_months', 'speed_limit_mph']

X, y = df[cat_cols + num_cols], df['problem']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42)

model = Pipeline([
    ('prep', ColumnTransformer([
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols),
        ('num', StandardScaler(), num_cols)])),
    ('clf', LogisticRegression(max_iter=1000, class_weight='balanced'))])
model.fit(X_train, y_train)
pred = model.predict(X_test)

print(f"Train: {len(X_train)}  Test: {len(X_test)}  (75/25 split, stratified)")
print(f"Accuracy : {accuracy_score(y_test, pred):.3f}")
print(f"Precision: {precision_score(y_test, pred):.3f}")
print(f"Recall   : {recall_score(y_test, pred):.3f}")
print(f"Baseline accuracy (always 'no problem'): {1 - y_test.mean():.3f}\n")
print(classification_report(y_test, pred, target_names=['No problem', 'Problem']))

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
ConfusionMatrixDisplay.from_predictions(y_test, pred, display_labels=['No problem', 'Problem'],
                                        cmap='Blues', ax=axes[0])
axes[0].set_title('Confusion matrix (test set)')
names = model.named_steps['prep'].get_feature_names_out()
coefs = pd.Series(model.named_steps['clf'].coef_[0], index=names)
top = coefs.reindex(coefs.abs().sort_values().index).tail(12)
axes[1].barh(top.index.str.replace(r'^(cat|num)__', '', regex=True), top.values,
             color=['tab:red' if v > 0 else 'tab:green' for v in top.values])
axes[1].axvline(0, c='k', lw=1)
axes[1].set_title('Top 12 factors (red = raises risk, green = lowers risk)')
plt.tight_layout(); plt.show()
```

```python
from sklearn.metrics import roc_auc_score

def evaluate(extra_num, label):
    cols_num = num_cols + extra_num
    Xr = df[cat_cols + cols_num].copy()
    if 'unexpected_obstacle' in extra_num:
        Xr['unexpected_obstacle'] = (Xr['unexpected_obstacle'] == 'Yes').astype(int)
    Xtr, Xte, ytr, yte = train_test_split(Xr, y, test_size=0.25, stratify=y, random_state=42)
    m = Pipeline([('prep', ColumnTransformer([
                      ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols),
                      ('num', StandardScaler(), cols_num)])),
                  ('clf', LogisticRegression(max_iter=1000, class_weight='balanced'))]).fit(Xtr, ytr)
    p = m.predict(Xte)
    return pd.Series({'accuracy': accuracy_score(yte, p), 'precision': precision_score(yte, p),
                      'recall': recall_score(yte, p),
                      'ROC AUC': roc_auc_score(yte, m.predict_proba(Xte)[:, 1])}, name=label)

results = pd.DataFrame([
    evaluate([], 'Pre-journey only'),
    evaluate(['sensor_confidence_pct', 'unexpected_obstacle'], '+ Real-time sensors')]).round(3)
print(results)
results.T.plot.bar(rot=0, figsize=(8, 4.5), title='Pre-journey vs real-time model (test set)')
plt.ylabel('Score'); plt.ylim(0, 1); plt.tight_layout(); plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree, export_text
from sklearn.model_selection import cross_val_score, StratifiedKFold

# Same pre-journey features and same split as the logistic model
X, y = df[cat_cols + num_cols], df['problem']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42)

def tree_pipeline(depth):
    return Pipeline([
        ('prep', ColumnTransformer([
            ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols),
            ('num', 'passthrough', num_cols)])),          # trees don't need scaling
        ('clf', DecisionTreeClassifier(max_depth=depth, min_samples_leaf=30,
                                       class_weight='balanced', random_state=42))])

# 1. Choose depth with cross-validation on the TRAINING set only (test set stays untouched)
cv = StratifiedKFold(5, shuffle=True, random_state=42)
depth_scores = pd.Series({d: cross_val_score(tree_pipeline(d), X_train, y_train, cv=cv,
                                              scoring='roc_auc').mean()
                          for d in range(1, 9)}, name='CV ROC AUC')
print(depth_scores.round(3), '\n')
best_depth = int(depth_scores.idxmax())
depth = min(best_depth, 4)        # cap at 4 so the tree stays readable
print(f"Best CV depth: {best_depth} -> using depth {depth}\n")

# 2. Fit and evaluate on the test set
tree = tree_pipeline(depth).fit(X_train, y_train)
pred_t = tree.predict(X_test)
print(classification_report(y_test, pred_t, target_names=['No problem', 'Problem']))

# 3. Compare with logistic regression
comparison = pd.DataFrame({
    'Logistic regression': [accuracy_score(y_test, pred), precision_score(y_test, pred),
                            recall_score(y_test, pred),
                            roc_auc_score(y_test, model.predict_proba(X_test)[:, 1])],
    f'Decision tree (depth {depth})': [accuracy_score(y_test, pred_t), precision_score(y_test, pred_t),
                                       recall_score(y_test, pred_t),
                                       roc_auc_score(y_test, tree.predict_proba(X_test)[:, 1])]},
    index=['accuracy', 'precision', 'recall', 'ROC AUC']).round(3)
print(comparison, '\n')

# 4. Charts: confusion matrix, comparison, the tree itself
names = [n.replace('cat__', '').replace('num__', '')
         for n in tree.named_steps['prep'].get_feature_names_out()]
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
ConfusionMatrixDisplay.from_predictions(y_test, pred_t, display_labels=['No problem', 'Problem'],
                                        cmap='Greens', ax=axes[0])
axes[0].set_title(f'Decision tree (depth {depth}) - test set')
comparison.plot.bar(ax=axes[1], rot=0, ylim=(0, 1), title='Logistic regression vs decision tree')
plt.tight_layout(); plt.show()

plt.figure(figsize=(22, 9))
plot_tree(tree.named_steps['clf'], feature_names=names, class_names=['No problem', 'Problem'],
          filled=True, rounded=True, proportion=True, fontsize=9)
plt.title('Decision tree rules (pre-journey factors)'); plt.show()

print(export_text(tree.named_steps['clf'], feature_names=names))   # the rules as text

imp = pd.Series(tree.named_steps['clf'].feature_importances_, index=names)
imp[imp > 0].sort_values().plot.barh(figsize=(8, 5), title='Decision tree feature importance')
plt.tight_layout(); plt.show()
```
